**Exercise:** Make a plot of the displacement vs. time of a particle undergoing simple harmonic motion with damping, where $$x(t) = A e^{- b t} \cos( \omega t + \phi ),$$ using $A = 3~{\rm cm}$, $b = 0.3~{\rm s}^{-1}$, $\omega = 1.5~{\rm rad/s}$, $\phi = \pi / 3$.  (Note that $\pi$ can be represented as ``np.pi`` and the cosine function is ``np.cos( )``.)  

Make sure that your variables are approrpriately and descriptively named. Finally, make sure that your graph looks nice, and is well labeled.

```python
def x_t(t):
    A = 3
    b = 0.3
    w = 0.75
    phi = np.pi/3
    return A * np.exp(-b * t)*np.cos(w*t+phi)
time = np.linspace(-1,15,100)
displacement= np.zeros(len(time))
for i in range(len(time)):
    displacement[i] = x_t(time[i])
plt.figure()
plt.plot(time,displacement)
plt.xlabel("Time (s)")
plt.ylabel("Displacement (cm)")
plt.title("Plot Harmonic emotion of a particle: Displacement vs Time")
plt.show()
```

**Exercise:** Create a parametric plot, where the points $(x, y) = (\cos \theta,\,\sin \theta)$ for $0 \leq \theta \leq 2\pi$.

```python
def variable_x_y(theta):
    return np.cos(theta), np.sin(theta)

theta = np.linspace (0, 2*np.pi, 100) 
x = np.zeros(len(theta))
y = np.zeros(len(theta))  

for i in range (len(theta)):
             x[i], y[i] = variable_x_y(theta[i])
plt.figure()
plt.figure(figsize = (6,6))
plt.plot( x, y)
plt.xlabel("cosine theta value")
plt.ylabel("sine theta value")
plt.title("Parametric Function (cosine, sine)")
plt.show
```

**Exercise:**  The file ``Lab2.csv``, has three columns: (0) is time (in seconds); (1) is position (in meters); (2) is velocity (in meters per second) for a $m = 2~{\rm kg}$ projectile falling under the influence of gravity and air resistance.  

Create two plots: **(1)** a plot of position vs. time (y vs. t); and **(2)** a plot of energy vs. time and momentum vs. time (this plot should include a legend so we know which curve is which).

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
m = 2  
g = 9.81  

def calculate_energy_momentum( v, y):
    E = 0.5 * m * v**2 + m * g * y 
    return E

def calculate_momentum(v):
    p = m * v
    return p

y = np.arange(1, 10, 1) 
v = np.arange(1, 10, 1)  

t = np.linspace(0, 10, len(y))
E = np.zeros(len(y))
p = np.zeros(len(y))

for i in range(len(y)):
    E[i]= calculate_energy_momentum (y[i],v[i])
    p[i] = calculate_momentum(v[i])

# Plot Position vs Time
plt.figure(figsize=(8, 6))
plt.plot(t, y, color='b', marker='^', label='Position (m)')  
plt.xlabel("Time (s)")
plt.ylabel("Position (m)")
plt.title("Position vs. Time")
plt.legend(loc='upper left')
plt.grid()
plt.show()

# Plot Energy vs time and Momentum vs Time
plt.figure(figsize=(8, 6))
plt.plot(t, E, color='g', marker='s', label='Energy (J)')
plt.plot(t, p, color='r', marker='o', label='Momentum (kg*m/s)')
plt.xlabel("Time (s)")
plt.ylabel("Energy and Momentum")
plt.title("Energy vs Time and Momentum vs. Time")
plt.legend(loc='upper left')
plt.grid()
plt.show()
plt.show()

```
