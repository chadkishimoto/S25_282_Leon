```python
import numpy as np
import matplotlib.pyplot as plt
```

**Exercise:** Create a heat map plot of the function
$$ f(x, y) = e^{- (x^2 + y^2)}$$
Using:
- 50 points in x from -2 to +2
- 200 points in y from -4 to +4

```python
x = np.linspace(-2, 2, 50) 
y = np.linspace(-4, 4, 200)
my_matrix = np.zeros((len(y), len(x)))  

def Map_function(x, y):
    return np.exp(-(x**2 + y**2))

for i in range(len(y)):
    for j in range(len(x)):
        my_matrix[i, j] = Map_function(x[j], y[i])
        
plt.figure(figsize=(8,8))
plt.imshow(my_matrix,extent=[-2,2,-4,4]) # new matrix
plt.colorbar()
plt.show()
```

**Exercise:**  Plot a heat map of the function
$$ g(x, y) = e^{- x^2} e^{- y^2 / 4}$$
using the same x and y arrays.

```python
x = np.linspace(-2, 2, 50) 
y = np.linspace(-4, 4, 200)
my_matrix = np.zeros((len(y), len(x)))  

def heat_function(x, y):
    return np.exp(-(x**2 + y**2 / 4))

for i in range(len(y)):
    for j in range(len(x)):
        my_matrix[i, j] = heat_function(x[j], y[i])
        
plt.figure(figsize=(8,8))
plt.imshow(my_matrix,extent=[-2,2,-4,4]) # new matrix
plt.colorbar()
plt.show()
```

**Exercise:** Create a contour plot of $g(x,y)$ with the same x and y arrays.
- Your contours should be at 75%, 50%, 25% of the maximum value
- Your contours should be labeled with their values
- You should include a symbol at the location of the maximum
- Your code should be completely self-contained, which means you haven't "hard coded" in any values.

```python
def g(x, y): 
    return np.exp(-x**2) * np.exp(-(y**2) / 4)

x = np.linspace(-2, 2, 50)
y = np.linspace(-4, 4, 200)

N = np.zeros((len(y), len(x)))
for i in range(len(y)): 
    for j in range(len(x)): 
        N[i, j] = g(x[j], y[i])

maximum_index = np.unravel_index(np.argmax(N), N.shape)

plt.figure(figsize=(8, 8))
CS = plt.contour(x, y, N, levels=[np.max(N) * 0.25, np.max(N) * 0.50, np.max(N) * 0.75], linewidths=2) 
plt.plot(x[maximum_index[1]], y[maximum_index[0]], '*', color='red', markersize=12)

plt.colorbar()
plt.title("Contour Plot of $g(x, y) = e^{-x^2} e^{-y^2/4}$")
plt.clabel(CS)
plt.show()
```
