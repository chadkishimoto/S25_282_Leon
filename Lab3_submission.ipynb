**Exercise:**  Use the data from ``Lab3bData.csv`` to create the following plots:
- A plot of $\chi^2$ vs $\lambda$
- A plot that shows the data points, with error bars, as well as the best fit model (using your best fit $\lambda$)

In the cell below, place the code that creates the $\chi^2$ vs $\lambda$ plot

```python

n_values, y_values, Uncertainty = np.loadtxt('Lab3bData.csv',delimiter=',',usecols=(0,1,2), unpack=True)
plt.errorbar(n_values, y_values, Uncertainty, fmt = 'o')
plt.xlabel("n values")
plt.ylabel(" y values (mm) ")
plt.title ( "Y VS N")
plt.show()


d = 0.125 / 1000 
L = 280 / 100 

def M(n, lambda_in_nm):
    lambda_m = lambda_in_nm * 1e-9  
    theta = np.arcsin(np.clip((n * lambda_m) / d, -1, 1))  
    return L * np.tan(theta) * 1000  

def Chi_squared(lambda_in_nm):
    chi_squared = np.sum(((y_values - M(n_values, lambda_in_nm)) ** 2) / (Uncertainty ** 2))
    return chi_squared

lambda_range = np.linspace(400, 700, 1000)
chi2_values = np.array([Chi_squared(lmbd) for lmbd in lambda_range])
plt.figure()
plt.plot(lambda_range, chi2_values, linewidth = 3)
plt.xlabel('Wavelength (nm)', size = 24)
plt.ylabel(r'$\chi^2$ Value', size = 24)
plt.title(r'$\chi^2$ Value$\, vs\, Wavelength$', fontsize=24)
```

In the cell below, place the code (that utilizes your other code above) to create the plots that include the data points with error bars; and the model curve for the wavelength.

```python
n_values, y_values, Uncertainty = np.loadtxt('Lab3bData.csv',delimiter=',',usecols=(0,1,2), unpack=True)

lambda_range = np.linspace(400, 700, 1000)
chi2_values = np.array([Chi_squared(lmbd) for lmbd in lambda_range])
np.where(chi2_values < 7)
lambda_range [np.where (chi2_values < 7.55)]
temp_array = (lambda_range [np.where (chi2_values < 7.55)])
 
n_fit = ((temp_array[-1] + temp_array[0])/2)
y_fit = ((temp_array[-1] - temp_array[0])/2)

plt.figure()
plt.plot(n_values, M(n_values,n_fit), linewidth = 2)
plt.xlabel('N values')
plt.ylabel('Y values (mm)')
plt.errorbar(n_values, y_values, yerr = Uncertainty, fmt='o')
plt.show()
```
